## 🎯 What is Tokenization?
Computers cannot read paragraphs like humans do. Before an algorithm can do Sentiment Analysis, Machine Translation, or Search, the raw text must be chopped up into small, discrete linguistic units called **Tokens**.

- **Sentence Tokenization:** Splitting a document or paragraph into individual sentences.
- **Word Tokenization:** Splitting a sentence into words, punctuation, numbers, and contractions.

### ❓ Why can't we just use Python's built-in `str.split()`?

Let's see what happens if we use basic `.split(' ')` vs NLTK's `word_tokenize`:

```python
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
from nltk.tokenize import sent_tokenize, word_tokenize

example_text = "Dr. Smith arrived at 8:30 a.m. in the U.S.A., but he didn't find his luggage!"

print("Raw Text:", example_text)
print("\n1. Python .split(' '):")
print(example_text.split(' '))

print("\n2. NLTK word_tokenize():")
print(word_tokenize(example_text))
```

### 💡 Notice the differences:
1. **Punctuation attached:** `.split(' ')` left `U.S.A.,` (with comma) and `luggage!` (with exclamation mark). A computer would think `"luggage!"` is a different word from `"luggage"`!
2. **Contractions:** NLTK split `"didn't"` into `['did', "n't"]` so the negation `n't` is isolated as its own grammatical token.
3. **Abbreviations:** NLTK's `sent_tokenize` understands that `"Dr."` and `"a.m."` are abbreviations and **not** the end of a sentence!

## 🎡 Real-World Lab: Tokenizing Disneyland Reviews Dataset

We load `DisneylandReviews.csv` using `pandas` with `encoding='latin1'` to handle special international characters safely.

```python
import os
import pandas as pd

csv_path = 'code/DisneylandReviews.csv' if os.path.exists('code/DisneylandReviews.csv') else 'DisneylandReviews.csv'
df = pd.read_csv(csv_path, encoding='latin1')

print(f"Dataset loaded with {len(df)} rows. Columns:")
print(list(df.columns))
print("\nPreview of the first row:")
display(df.head(2))
```

### 📝 Step 1: Extracting a Review

```python
# Grab the review text from row 0
review = str(df['Review_Text'][0])

print("Original Review:")
print("-" * 60)
print(review)
```

### ✂️ Step 2: Sentence Tokenization (`sent_tokenize`)
Splitting the customer review into individual sentences.

```python
sentences = sent_tokenize(review)

print(f"Found {len(sentences)} Sentences:\n")
for i, s in enumerate(sentences, start=1):
    print(f"[{i}] {s}")
```

### 🔍 Step 3: Word Tokenization (`word_tokenize`)
Splitting into individual words and punctuation tokens.

```python
words = word_tokenize(review)

print(f"Total Tokens: {len(words)}\n")
print("Sample of first 25 tokens:")
print(words[:25])
```

### 📊 Step 4: Token Statistics & Frequency Distribution
Let's see the most frequently used words in this review.

```python
from collections import Counter

# Filter to only alphabetic words (exclude punctuation) and lowercase them
cleaned_words = [w.lower() for w in words if w.isalpha()]

word_freq = Counter(cleaned_words)
print("Top 10 Most Frequent Words:")
print("-" * 30)
for word, count in word_freq.most_common(10):
    print(f"{word:<12} : {count}")
```

## 🎓 Summary of Key Functions

| NLTK Function | Purpose | Input $\rightarrow$ Output |
|---|---|---|
| `nltk.tokenize.sent_tokenize(text)` | Splits paragraph into sentences | `str` $\rightarrow$ `list[str]` |
| `nltk.tokenize.word_tokenize(text)` | Splits text into words & punctuation | `str` $\rightarrow$ `list[str]` |
| `text.isalpha()` | Checks if token is a word (not a dot or comma) | `bool` |
